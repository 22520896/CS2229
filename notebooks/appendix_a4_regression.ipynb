# Phụ lục A.4 · Bốn hồi quy kiểm chứng

Ước lượng hệ số cho làm trơn, bỏ hoặc thêm đặc trưng và xoay. Mỗi hồi quy dùng điểm test chuẩn hóa của cấu hình lấy theo validation. Mô hình có hiệu ứng cố định theo bộ dữ liệu; các hàng hiệu ứng này được ẩn như trong paper. Seed 0 áp dụng cho đối chiếu độc lập với R. Cột estimate của tác giả trích từ CSV lịch sử đã làm tròn, để đọc cạnh kết quả chạy lại. Mốc hồi quy có thể ảnh hưởng đến hệ số chặn.

```python
from pathlib import Path
import sys

root = next(p for p in (Path.cwd(), *Path.cwd().parents) if (p / "src/analysis.py").is_file())
sys.path.insert(0, str(root))
from IPython.display import Image, Markdown, display
import pandas as pd

from src.analysis import scores
from src.appendix_a4 import analyze, parity
from src.appendix_notebooks import REGRESSIONS
coefficients = analyze(scores("ablations"))
reference = pd.read_csv(root / "evidence/a4_r_reference.csv", float_precision="round_trip")
display(parity(coefficients, reference))
published = pd.read_csv(root / "evidence/a4_author_tables.csv")
coefficients.to_csv(root / "evidence/a4_coefficients.csv", index=False)
```

```python
for analysis, title in REGRESSIONS.items():
    rows = coefficients[coefficients.analysis.eq(analysis)].merge(
        published.loc[published.analysis.eq(analysis), ["term", "estimate"]],
        on="term", suffixes=("", "_author"), validate="one_to_one")
    display(Markdown(f"### {title} · {rows.nobs.iloc[0]:,} quan sát"))
    for column, digits in {"estimate": 6, "estimate_author": 3, "std_error": 6, "statistic": 4}.items():
        rows[column] = rows[column].map(lambda value: f"{value:.{digits}f}")
    rows["p_value"] = rows["p_value"].map(lambda value: f"{value:.2e}")
    display(rows[["term", "estimate", "estimate_author", "std_error", "statistic", "p_value"]])
```

# Phụ lục A.1 và A.3 · Danh mục dữ liệu và HPO

Danh mục ghi 57 thiết lập theo nhóm bài toán và loại đặc trưng. Một bộ dữ liệu có thể xuất hiện trong nhiều thiết lập. Paper mô tả 45 bộ dữ liệu; giá trị này không phải số hàng của danh mục.

```python
from pathlib import Path
import sys

root = next(p for p in (Path.cwd(), *Path.cwd().parents) if (p / "src/analysis.py").is_file())
sys.path.insert(0, str(root))
from IPython.display import Image, Markdown, display
import pandas as pd

datasets = pd.read_csv(root / "evidence/appendix_datasets.csv")
suites = {337: "Số · Phân loại", 336: "Số · Hồi quy",
          334: "Có biến phân loại · Phân loại", 335: "Có biến phân loại · Hồi quy"}
for suite, label in suites.items():
    rows = datasets[datasets.suite.eq(suite)]
    display(Markdown(f"### {label} · {len(rows)} thiết lập"))
    display(rows[["paper_name", "processed_dataset_id", "task", "rows", "features"]].reset_index(drop=True))
```

## Không gian tìm kiếm

Bảng 1–8 ghi phân phối lấy mẫu hoặc giá trị cố định. Cột mặc định để trống ở các bảng cây khi paper không in riêng. Riêng SAINT, bảng in mặc định batch 512 và tập random [128, 256]; cả hai được giữ nguyên.

```python
hpo = pd.read_csv(root / "evidence/appendix_hpo.csv", keep_default_na=False)
for number, rows in hpo.groupby("table"):
    display(Markdown(f"### Bảng {number} · {rows.model.iloc[0]}"))
    display(rows[["parameter", "distribution", "default"]].reset_index(drop=True))
```
